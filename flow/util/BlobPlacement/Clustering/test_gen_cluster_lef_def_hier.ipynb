```python
csv_file1 = '/home/fetzfs_projects/PlacementCluster/bodhi/BlobPlacement/test_area/ariane_NG45/ariane_cluster_leiden.csv'
design = 'ariane'
design_dir = '/home/fetzfs_projects/PlacementCluster/sakundu/BlobPlacement/blob_runs/group_defs'
csv_file2 = f'{design_dir}/{design}_ng45_clusters.rpt'
df1 = pd.read_csv(csv_file1)
df1 = df1[['Name', 'Cluster_id']]
df2 = pd.read_csv(csv_file2)
l1 = len(df1['Cluster_id'].drop_duplicates())
l2 = len(df2['Cluster_id'].drop_duplicates())
print(f'Number of clusters in Leiden {l1}')
print(f'Number of clusters in Logical hierarchy {l2}')
```

```python
df = df1.merge(df2, on='Name', how='inner')
df = df.sort_values(by=['Cluster_id_x', 'Cluster_id_y'])
cluster_id_df = df[['Cluster_id_x', 'Cluster_id_y']].drop_duplicates().reset_index(drop=True)
number_cluster = len(cluster_id_df)
cluster_id_df['Cluster_id'] = range(number_cluster)
df = df.merge(cluster_id_df, on=['Cluster_id_x', 'Cluster_id_y'], how='inner')
df = df.drop(columns=['Cluster_id_x', 'Cluster_id_y'])
df.to_csv(f'{design_dir}/{design}_overlay_cluster.csv', index=False)
```
